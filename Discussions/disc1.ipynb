Q1: Race
y=4
x=6
It doesnt work because the tortoise catches it

```python
def race(x, y):
    """The tortoise always walks x feet per minute, while the hare repeatedly
    runs y feet per minute for 5 minutes, then rests for 5 minutes. Return how
    many minutes pass until the tortoise first catches up to the hare.

    The tortoise catches up to the hare when it has gone a greater distance
    than the hare.

    These examples work correctly:

    >>> race(5, 7)  # After 7 minutes, both have gone 35 steps
    7
    >>> race(2, 4) # After 10 minutes, both have gone 20 steps
    10
    """
    assert y > x and y <= 2 * x, 'the hare must be faster but at most 2x faster'
    tortoise, hare, minutes = 0, 0, 0
    while (minutes == 0) or (tortoise - hare):
        tortoise += x
        if minutes % 10 < 5:
            hare += y
        minutes += 1
    return minutes

race(4,6)
```

Q3 Is Prime?
Starts at smallest possible multiple,checks mod==0, if yes return true else multiple increments  by 1 and reloop

```python
def is_prime(n):
    """
    >>> is_prime(10)
    False
    >>> is_prime(7)
    True
    >>> is_prime(1) # one is not a prime number!!
    False
    """
    i=2
    while i<n:
        if n%i==0:
            return False
        i+=1
```

Repeating

```python
def repeating(t, n):
    """Return whether t digits repeat to form positive integer n.

    >>> repeating(1, 6161)
    False
    >>> repeating(2, 6161)  # repeats 61 (2 digits)
    True
    >>> repeating(3, 6161)
    False
    >>> repeating(4, 6161)  # repeats 6161 (4 digits)
    True
    >>> repeating(5, 6161)  # there are only 4 digits
    False
    """
    if pow(10, t-1) > n:  # make sure n has at least t digits
        return False
    end = t
    while n:
        if n % pow(10, t) != end:
           return False
        n//=10
    return True
```
